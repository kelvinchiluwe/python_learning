```python
class vehicle:
    def __init__(self,brand,rental_price):
        self.brand=brand
        self.rental_price=rental_price
    def rent(self):
        print(f'your payment is successful {self.rental_price} and you have borrowed {self.brand}')

class car(vehicle):
    def __init__(self,brand,rental_price,fuel_efficiency):
        self.fuel_efficiency=fuel_efficiency
        super().__init__(brand,rental_price)
    def rent(self):
        super().rent()
        print(f'fuel efficiency{self.fuel_efficiency}')
```

```python
dd=car('dd',1234,'25 MPG')
dd.rent()
```
